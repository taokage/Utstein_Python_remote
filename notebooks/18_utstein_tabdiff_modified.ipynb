# 18：df_modified全体のTabDiff学習

01で保存した `data/processed/utstein_modified.parquet` を読み込み、指定症例数を抽出し、90%を学習・10%をホールドアウトとして1モデルを学習します。`MAX_TRAIN_ROWS = "all"` で全件を使います。

学習曲線・経過時間を表示し、通常モデルとEMAモデルの両方を保存します。13種類の連続学習は21を使用してください。

前処理・モデルの保存先は `artifacts/18_tabdiff/実行時刻/` です。19で生成し、20で検証できます。データセット一覧JSONは不要です。

01で条件に応じて欠損化した列は、数値では欠損フラグ、カテゴリーでは欠損カテゴリーとして学習します。生成結果がその条件を必ず守ることを保証するものではありません。


```python
from pathlib import Path
from datetime import datetime
import os, sys, json, subprocess, hashlib, time
import numpy as np
import pandas as pd
from IPython.display import display, HTML
import matplotlib.pyplot as plt

PROJECT = Path.cwd().resolve()
if PROJECT.name == "notebooks":
    PROJECT = PROJECT.parent
# df_modified全体を1つのTabDiffモデルとして学習します。
DATASET_NAME = "df_modified"
PROCESSED_DIR = PROJECT / "data/processed"
INPUT = PROCESSED_DIR / "utstein_modified.parquet"
TABDIFF_REPO = PROJECT / "external/TabDiff"  # 既存の公式コードがあれば変更
TABDIFF_PYTHON = Path(sys.executable)       # TabDiff環境のPython
# ===== 学習条件：実行前に空欄へ数値を入力してください =====
# 症例数："all"で対象サブセットの全件。整数なら指定件数を抽出して90%/10%分割します。
MAX_TRAIN_ROWS = "all"          # 例: 100_000（"all"なら全症例）
EPOCHS = "200"                  # 例: 200
TRAIN_BATCH_SIZE = "4096"        # 例: 1024（GPUメモリ不足なら小さくします）
SAMPLE_BATCH_SIZE = "512"       # 例: 512（学習後の生成用）
DIFFUSION_STEPS = "50"         # 例: 50（拡散の時間ステップ数）
CHECKPOINT_SELECTION = "ema"    # "ema" または "normal"（最終エポックのEMA/通常重み）
N_GENERATE = 10_000           # 19で生成する件数の初期値

GPU = 0                       # CPUなら-1（大規模データは非常に時間がかかります）
LOSS_EVAL_EVERY = 5          # 固定評価データで損失を測る間隔（epoch）
LOSS_EVAL_ROWS = 2048        # 学習・ホールドアウトそれぞれの評価上限（全件評価はしません）
LOSS_EVAL_BATCH_SIZE = 512  # 評価用バッチサイズ
SEED = 42
EMA_DECAY = 0.997             # 各バッチでEMAを更新


def _required_positive_int(name, value):
    """空欄を許さず、正の整数として学習設定を検証します。"""
    if value == "" or value is None:
        raise ValueError(f"{name} が空欄です。上の学習条件に正の整数を入力してください。")
    try:
        number = int(value)
    except (TypeError, ValueError) as exc:
        raise ValueError(f"{name} は整数で入力してください: {value!r}") from exc
    if number <= 0:
        raise ValueError(f"{name} は1以上にしてください。")
    return number

USE_ALL_ROWS = str(MAX_TRAIN_ROWS).strip().lower() == "all"
if not USE_ALL_ROWS:
    MAX_TRAIN_ROWS = _required_positive_int("症例数 MAX_TRAIN_ROWS", MAX_TRAIN_ROWS)
EPOCHS = _required_positive_int("エポック数 EPOCHS", EPOCHS)
TRAIN_BATCH_SIZE = _required_positive_int("学習バッチサイズ TRAIN_BATCH_SIZE", TRAIN_BATCH_SIZE)
SAMPLE_BATCH_SIZE = _required_positive_int("生成バッチサイズ SAMPLE_BATCH_SIZE", SAMPLE_BATCH_SIZE)
DIFFUSION_STEPS = _required_positive_int("拡散ステップ数 DIFFUSION_STEPS", DIFFUSION_STEPS)
N_GENERATE = _required_positive_int("生成件数 N_GENERATE", N_GENERATE)
LOSS_EVAL_EVERY = _required_positive_int("LOSS_EVAL_EVERY", LOSS_EVAL_EVERY)
LOSS_EVAL_ROWS = _required_positive_int("LOSS_EVAL_ROWS", LOSS_EVAL_ROWS)
LOSS_EVAL_BATCH_SIZE = _required_positive_int("LOSS_EVAL_BATCH_SIZE", LOSS_EVAL_BATCH_SIZE)
CHECKPOINT_SELECTION = str(CHECKPOINT_SELECTION).strip().lower()
if CHECKPOINT_SELECTION not in {"ema", "normal"}:
    raise ValueError('CHECKPOINT_SELECTION は "ema" または "normal" を入力してください。')

EXCLUDE_COLUMNS = ["ID", "id", "患者ID", "患者番号", "症例ID"]
RUN_ID = datetime.now().strftime("%Y%m%d_%H%M%S_%f")
DATA_NAME = "utstein_" + DATASET_NAME.removeprefix("df_") + "_" + RUN_ID
EXP_NAME = "unconditional"
OUT = PROJECT / "artifacts/18_tabdiff" / RUN_ID
assert INPUT.is_file(), f"入力がありません: {INPUT}"
assert (TABDIFF_REPO / "main.py").is_file(), f"公式TabDiffのパスを指定してください: {TABDIFF_REPO}"
OUT.mkdir(parents=True, exist_ok=False)
print("入力:", INPUT, "\n学習出力:", OUT)
def sha256_file(path):
    h = hashlib.sha256()
    with Path(path).open('rb') as f:
        for block in iter(lambda: f.read(1024**2), b''):
            h.update(block)
    return h.hexdigest()

REQUESTED_TRAIN_ROWS = MAX_TRAIN_ROWS
REQUESTED_USE_ALL_ROWS = USE_ALL_ROWS
```

```python
# 損失は小さいほどよい指標ですが、生成データの正確さを直接表す「正解率」ではありません。
def make_learning_figure(history):
    fig, axes = plt.subplots(2, 2, figsize=(13, 8))
    online = [("loss/total_loss", "Train online"), ("ema_loss/total_loss", "EMA train evaluation")]
    panels = [
        (axes[0, 0], "Online training / full-train EMA loss", online),
        (axes[0, 1], "Fixed subset: total loss", [(f"eval/{s}/{m}/total", f"{s} {m}") for s in ("train", "holdout") for m in ("normal", "ema")]),
        (axes[1, 0], "Fixed subset: numerical loss", [(f"eval/{s}/{m}/continuous", f"{s} {m}") for s in ("train", "holdout") for m in ("normal", "ema")]),
        (axes[1, 1], "Fixed subset: categorical loss", [(f"eval/{s}/{m}/categorical", f"{s} {m}") for s in ("train", "holdout") for m in ("normal", "ema")]),
    ]
    for ax, title, lines in panels:
        for key, label in lines:
            if key in history:
                values = pd.to_numeric(history[key], errors="coerce")
                good = values.notna()
                if good.any():
                    ax.plot(history.loc[good, "epoch"], values[good], label=label,
                            linestyle="--" if "holdout" in label else "-", marker=".", markersize=4)
        ax.set(title=title, xlabel="Epoch", ylabel="Loss (lower is better)")
        ax.grid(alpha=.25)
        if ax.lines:
            ax.legend(fontsize=8)
    fig.suptitle("TabDiff learning curves: loss is not synthetic-data accuracy")
    fig.tight_layout()
    return fig

# 標準出力・進捗バー・警告はログファイルへ保存します。
# 画面の表示は同じ場所を更新し、ログ行を大量に増やしません。
from collections import deque

def format_duration(seconds):
    seconds = max(0, int(seconds))
    h, rem = divmod(seconds, 3600)
    m, sec = divmod(rem, 60)
    return f"{h:02d}:{m:02d}:{sec:02d}"

def run_command(args, log_name, show_timing=False):
    env = os.environ.copy()
    env.update(WANDB_MODE="disabled", PYTHONUNBUFFERED="1", PYTHONHASHSEED=str(SEED))
    log_path = OUT / log_name
    progress_path = OUT / "training_progress.json"
    if show_timing:
        progress_path.unlink(missing_ok=True)
        (OUT / "training_history.jsonl").unlink(missing_ok=True)
    started = time.monotonic()
    handle = None
    graph_handle = None
    last_graph_epoch = None
    def update_graph():
        nonlocal graph_handle, last_graph_epoch
        history_path = OUT / "training_history.jsonl"
        if not history_path.is_file():
            return
        # 末尾が書き込み途中でも、完了したJSON行だけを利用します。
        rows = []
        for line in history_path.read_text(encoding="utf-8").splitlines():
            try:
                rows.append(json.loads(line))
            except ValueError:
                continue
        if not rows or rows[-1]["epoch"] == last_graph_epoch:
            return
        history = pd.DataFrame(rows)
        last_graph_epoch = int(history["epoch"].iloc[-1])
        fig = make_learning_figure(history)
        if graph_handle is None:
            graph_handle = display(fig, display_id=True)
        else:
            graph_handle.update(fig)
        fig.savefig(OUT / "training_learning_curves.png", dpi=160, bbox_inches="tight")
        history.to_csv(OUT / "training_history.csv", index=False, encoding="utf-8-sig")
        plt.close(fig)
    def update_display(finished=False):
        nonlocal handle
        elapsed = time.monotonic() - started
        estimate = None
        if finished:
            estimate = elapsed
        elif progress_path.is_file():
            try:
                progress = json.loads(progress_path.read_text(encoding="utf-8"))
                done = progress.get("completed_epochs", 0)
                if done > 0:
                    # 起動・前処理時間も含む総経過時間を完了epoch数で外挿。
                    estimate = elapsed + progress["training_elapsed"] / done * max(0, EPOCHS-done)
            except (OSError, ValueError, KeyError):
                pass
        text = ("現在の経過時間：" + format_duration(elapsed)
                + "\n予想総所要時間：" + (format_duration(estimate) if estimate is not None else "推定中（最初の1エポック完了後に表示）"))
        if handle is None:
            handle = display(text, display_id=True)
        else:
            handle.update(text)
        update_graph()
    with log_path.open("w", encoding="utf-8") as log:
        proc = subprocess.Popen(list(map(str, args)), cwd=TABDIFF_REPO, env=env,
                                stdout=log, stderr=subprocess.STDOUT)
        try:
            if show_timing:
                update_display()
            last_update = time.monotonic()
            while proc.poll() is None:
                time.sleep(1)
                if show_timing and time.monotonic()-last_update >= 5:
                    update_display()
                    last_update = time.monotonic()
        except KeyboardInterrupt:
            proc.terminate()
            try:
                proc.wait(timeout=10)
            except subprocess.TimeoutExpired:
                proc.kill()
                proc.wait()
            print(f"中断しました。ログ: {log_path}")
            raise
    if proc.returncode:
        with log_path.open(encoding="utf-8", errors="replace") as f:
            print("".join(deque(f, maxlen=80)))
        raise RuntimeError(f"処理失敗。ログ: {log_path}")
    if show_timing:
        update_display(finished=True)

# GPU環境確認の詳細もenvironment.logへ保存します。
run_command([TABDIFF_PYTHON, "-c",
    "import torch,pandas,numpy,sklearn,sdmetrics,wandb,tomli,tomli_w; "
    "print('torch',torch.__version__);print('CUDA',torch.cuda.is_available()); "
    "print(torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU')"], "environment.log")
commit = subprocess.check_output(["git", "rev-parse", "HEAD"], cwd=TABDIFF_REPO, text=True).strip()
```

## df_modifiedの前処理


```python
real = pd.read_parquet(INPUT, engine="pyarrow")
if not isinstance(real, pd.DataFrame) or not real.columns.is_unique:
    raise ValueError("入力は列名が一意のDataFrameにしてください。")
real = real.drop(columns=[c for c in EXCLUDE_COLUMNS if c in real.columns]).copy()
# df_modifiedはサブセット一覧JSONに依存せず、Parquetの全列を学習します。
if "心臓マッサージ" in real.columns:
    raise ValueError("旧版のdf_modifiedです。新版01でParquetを保存し直してください。")
if real.columns[0] != "覚知ウツタイン":
    raise ValueError("覚知ウツタインを先頭列にした新版01の出力を使用してください。")
dataset_entry = {"dataset":DATASET_NAME,"file":INPUT.name,"rows":len(real),
                 "columns":real.columns.tolist(),"conditions":[],
                 "aed_positive_value":"1"}
INPUT_SHA256 = sha256_file(INPUT)
INPUT_ROWS = len(real)
if USE_ALL_ROWS:
    MAX_TRAIN_ROWS = INPUT_ROWS
print(f"対象: {DATASET_NAME} / 元データ {INPUT_ROWS:,}件 / 抽出予定 {MAX_TRAIN_ROWS:,}件")

if MAX_TRAIN_ROWS > len(real):
    raise ValueError(f"指定症例数 {MAX_TRAIN_ROWS:,} が元データ件数 {len(real):,} を超えています。")
real = real.sample(n=MAX_TRAIN_ROWS, random_state=SEED)
real = real.reset_index(drop=True)
if len(real) < 20:
    raise ValueError("少なくとも20行必要です。")
rng = np.random.default_rng(SEED)
indices = rng.permutation(len(real))
cut = int(len(real) * .9)
train_idx, test_idx = indices[:cut], indices[cut:]
schema, encoded = {}, pd.DataFrame(index=real.index)
constants = {}
num_names, cat_names = [], []
for i, column in enumerate(real.columns):
    series = real[column]
    key = f"f{i:03d}"
    # 日付のみのobject列にも対応
    is_date = pd.api.types.is_datetime64_any_dtype(series) or column in ["発生年月日ウツタイン", "覚知ウツタイン"]
    if series.nunique(dropna=False) <= 1:
        constants[column] = series.iloc[0]
        continue
    spec = {"key": key, "dtype": series.dtype}
    if is_date or (pd.api.types.is_numeric_dtype(series) and not isinstance(series.dtype, pd.CategoricalDtype)):
        if is_date:
            dt = pd.to_datetime(series, errors="coerce")
            values = (dt - pd.Timestamp("1970-01-01")).dt.total_seconds() / 86400
            spec["kind"] = "datetime"
        else:
            values = pd.to_numeric(series, errors="coerce").astype(float)
            spec["kind"] = "numeric"
        values = values.replace([np.inf, -np.inf], np.nan)
        observed = values.loc[train_idx].dropna()
        if observed.empty:
            raise ValueError(f"学習側に有効な数値がありません: {column}")
        spec.update(median=float(observed.median()), minimum=float(observed.min()),
                    maximum=float(observed.max()), integer=bool((observed % 1 == 0).all()))
        encoded[key] = values.fillna(spec["median"])
        num_names.append(key)
        if values.isna().any():
            spec["mask"] = key + "_missing"
            encoded[spec["mask"]] = np.where(values.isna(), "missing_yes", "missing_no")
            cat_names.append(spec["mask"])
    else:
        spec["kind"] = "category"
        labels = series.astype("string")
        unique = labels.dropna().unique().tolist()
        if len(unique) > 2000:
            raise ValueError(f"カテゴリーが多すぎます。ID等でないか確認してください: {column} ({len(unique)})")
        mapping = {value: f"label_{j:05d}" for j, value in enumerate(unique)}
        spec["decode"] = {v: k for k, v in mapping.items()}
        encoded[key] = labels.map(mapping).fillna("missing_category")
        cat_names.append(key)
    schema[column] = spec
if not num_names:
    raise ValueError("このノートブックでは少なくとも1つの可変数値列が必要です。")
# 公式実装のtarget欄に数値列1つを指定します。目的変数を条件固定せず全列を同時生成します。
target_key = num_names[0]
metadata = {"name": DATA_NAME, "task_type": "regression", "header": "infer",
    "column_names": encoded.columns.tolist(), "file_type": "csv",
    "num_col_idx": [encoded.columns.get_loc(c) for c in num_names if c != target_key],
    "cat_col_idx": [encoded.columns.get_loc(c) for c in cat_names],
    "target_col_idx": [encoded.columns.get_loc(target_key)],
    "data_path": f"data/{DATA_NAME}/{DATA_NAME}.csv",
    "test_path": f"data/{DATA_NAME}/holdout.csv", "val_path": None}
DATA_DIR = TABDIFF_REPO / "data" / DATA_NAME
DATA_DIR.mkdir(parents=True, exist_ok=False)
(TABDIFF_REPO / "data/Info").mkdir(parents=True, exist_ok=True)
encoded.iloc[train_idx].to_csv(DATA_DIR / f"{DATA_NAME}.csv", index=False)
encoded.iloc[test_idx].to_csv(DATA_DIR / "holdout.csv", index=False)
(TABDIFF_REPO / "data/Info" / f"{DATA_NAME}.json").write_text(json.dumps(metadata, indent=2), encoding="utf-8")
pd.to_pickle({"schema": schema, "constants": constants, "columns": real.columns.tolist(),
              "train_idx": train_idx, "test_idx": test_idx, "dtypes": real.dtypes.to_dict(), "dataset_name": DATASET_NAME, "dataset_conditions": dataset_entry["conditions"]}, OUT / "preprocessing.pkl")
display(pd.DataFrame({"列名": list(schema), "変換": [s["kind"] for s in schema.values()]}))
print("抽出症例数:", len(real), "学習:",len(train_idx),"ホールドアウト:",len(test_idx),"モデル入力列数:",encoded.shape[1])
```

```python
# 公式スクリプトでNumPy配列とinfo.jsonを生成します。
run_command([TABDIFF_PYTHON, "process_dataset.py", "--dataname", DATA_NAME], "preprocess.log")
# 同じ前処理情報・学習/ホールドアウト表の組み合わせを記録します。
training_assets = {"dataset_name": DATASET_NAME, "input_path": str(INPUT),
    "input_sha256": INPUT_SHA256, "input_rows": INPUT_ROWS,
    "dataset_conditions": dataset_entry["conditions"],
    "aed_positive_value": dataset_entry["aed_positive_value"],
    
    "preprocessing_sha256": sha256_file(OUT / "preprocessing.pkl"),
    "train_csv_sha256": sha256_file(DATA_DIR / f"{DATA_NAME}.csv"),
    "holdout_csv_sha256": sha256_file(DATA_DIR / "holdout.csv"),
    "ema_update": "per_batch", "ema_decay": EMA_DECAY,
    "selected_checkpoint": CHECKPOINT_SELECTION, "epochs": EPOCHS,
    "train_batch_size": TRAIN_BATCH_SIZE, "diffusion_steps": DIFFUSION_STEPS,
    "requested_rows": MAX_TRAIN_ROWS,
    "dataname": DATA_NAME, "exp_name": EXP_NAME,
}
(OUT / "training_assets.json").write_text(json.dumps(training_assets, indent=2), encoding="utf-8")
```
